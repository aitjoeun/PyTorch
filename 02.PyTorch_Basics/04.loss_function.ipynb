### Loss Function(손실함수)
: 모델의 예측값이 실제 정답과 얼마나 다른지 수치로 계산하는 함수<br>
  모델의 목표는 손실값을 줄이는 것

- 기본용어
    - Prediction : 모델이 예측한 값
    - Target : 정답
    - Loss : 예측값과 정답의 차이를 계산한 값

- 손실함수의 역할
    - 예측값과 정답의 차이를 수치로 표현
    - 모델이 얼마나 잘 예측했는지 평가할 수 있음
    - 자동 미분을 통해 weight와 bias를 업데이트할 때 사용할 기울기를 계산할 수 있음

#### 1. MSELoss(Mean Squared Error)
: 평균 제곱 오차를 계산하는 손실함수
예측값과 정답의 차이를 제곱한 뒤 평균을 계산함

- 특징
    - 오차를 제곱하므로 큰 오차에 더 큰 패널티를 부여함
    - 손실값이 0이상임
    - 예측값이 정답과 같으면 손실값은 0
    - 주로 회귀문제에서 사용

- 사용방법
loss_fn = mm.MSELoss()
loss = loss_fn(prediction, target)

기본설정 reduction = 'mean'은 모든 원소의 제곱오차의 평균
