#### ▶ PyTorch의 함수들

#### ▶ 난수 생성 함수
| 함수 | 분포 | 범위및 특징 |
| :---: | :---: | :---: |
| torch.rand() | 균등분포 | 0 ~ 1 |
| torch.randn() | 정규분포 | 평균 0, 표준편차 1 |
| torch.randint() | 이산균등분포 | 정수 난수 |

```python
import torch
```

```python
print(torch.rand(3, 3))
print(torch.randn(3, 3))
print(torch.randint(3, 10, (5,)))
print(torch.randint(3, 10, (2,3)))
```

#### ▶ PyTorch 주요 수학 함수

#### 1. 기본 수학 함수

| 함수 | 의미 | 예시 | 설명 |
| :---: | :---: | :---: | :--- |
| torch.abs(a) | 절대값 | abx(-3) -> 3 | 모든 원소의 절대값 반환 |
| torch.sqrt(a) | 제곱근 | sqrt(9) -> 3 | $\sqrt{x}$ |
| torch.square(a) | 제곱 | square(3) -> 9 | $x^2$ |
| torch.pow(a, n) | 거듭제곱| pow(2, 3) -> 8 | $x^n$ |
| torch.exp(a) | 지수함수 | exp(1) -> e | $e^x$ |
| torch.log(a) | 자연로그 | log(e) -> 1 | ln(x) |
| torch.log10(a) | 상용로그 | log10(10) -> 1 | $\log_{10}(x)$ |
| torch.log2(a) | 밑이 2인 로그 | log2(2) -> 1 | $\log_2(x)$ |


```python
a = torch.randn(3, 3)
print(a)

# 기본 수학 함수
print('절댓값 :\n', torch.abs(a))
print('제곱근 :\n', torch.sqrt(torch.abs(a)))
print('제곱 :\n', torch.square(a))
print('거듭제곱 : \n', torch.pow(a, 2))

# 지수함수
print('지수함수 : \n', torch.exp(a)) # 2.7183^0.5847
print('자연로그 : \n', torch.log(torch.abs(a))) # logeX
print('자연로그 : \n', torch.log(torch.exp(torch.tensor(1))))
print('상용로그 : \n', torch.log10(torch.tensor(10)))
print('밑이2인 로그 : \n', torch.log2(torch.tensor(2)))
```

#### 2. 반올림 / 올림 / 내림

| 함수 | 의미 | 예시 |
| :---: | :---: | :---: |
| torch.round(a) | 반올림 | round(1.6) -> 2 |
| torch.round(a, decimals=2) | 지정한 소수점까지 반올림 | 1.6275 -> 1.63 |
| torch.floor(a) | 내림 | 1.9 -> 1 |
| torch.ceil(a) | 올림 | 1.1 -> 2 |
| torch.trunc(a) | 소수점 버림  | 1.87 -> 1 |

```python
print(a)
print(torch.round(a))
print(torch.round(a, decimals=2))
print(torch.floor(a))
print(torch.ceil(a))
print(torch.trunc(a))
```

#### 3. 최솟값 / 최댓값

| 함수 | 의미 | 설명 |
| :---: | :---: | :---: |
| torch.min(a) | 최솟값 | 전체 원소 중 가장 작은값 |
| torch.max(a) | 최댓값 | 전체 원소 중 가장 큰값 |
| torch.argmin(a) | 최솟값의 인덱스 | 최솟값이 위치한 인덱스 반환 |
| torch.argmax(a) | 최댓값의 인덱스 | 최댓값이 위치한 인덱스 반환 |

```python
a = torch.randn(3, 4)
print(a)
```

```python
print('최댓값 :', torch.max(a))

# 옵션 dim : 각 차원에서 가장 큰값과 그값의 인덱스 반환
print(torch.max(a, dim=0)) # 열끼리
print(torch.max(a, dim=1)) # 행끼리
print('-' * 50)

# 옵션 keepdims : 최댓값을 반환한 후에도 텐서의 차원을 그대로 유지할 것인지 여부(기본값 : False)
print(torch.max(a, dim=0, keepdim=True))
print(torch.max(a, dim=1, keepdim=True))
print('-' * 50)

print(a)
print(torch.min(a))
print(torch.min(a, dim=0))
print(torch.min(a, dim=1))
print('-' * 50)

print(torch.argmax(a))
print(torch.argmax(a, dim=0))
print(torch.argmax(a, dim=1))
```
